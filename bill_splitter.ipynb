```python
# bill_splitter.py

def get_number_of_people():
    while True:
        try:
            count = int(input("Enter the number of people to split the bill: "))
            if count > 1:
                return count
            else:
                print("Enter at least 2 people.")
        except ValueError:
            print("Please enter a valid number.")

def get_bill_amount():
    while True:
        try:
            amount = float(input("Enter the total bill amount: ₹"))
            if amount >= 0:
                return amount
            else:
                print("Amount cannot be negative.")
        except ValueError:
            print("Please enter a valid amount.")

def get_names(count):
    names = []
    print("\nEnter the names of the people:")
    for i in range(count):
        name = input(f"Person {i+1} name: ").strip()
        names.append(name if name else f"Person{i+1}")
    return names

def split_bill(total_amount, names):
    share = round(total_amount / len(names), 2)
    print("\n💸 Bill Split Summary:\n")
    for name in names:
        print(f"{name} should pay: ₹{share}")

def main():
    print("🧾 Welcome to the Bill Splitter App!\n")
    people_count = get_number_of_people()
    total_amount = get_bill_amount()
    names = get_names(people_count)
    split_bill(total_amount, names)

# Run the app
if __name__ == "__main__":
    main()
```
